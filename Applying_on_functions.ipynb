2.	Calculate the factorial of a given number.

```python
#Calculate the factorial of a given number from the user input

def factorial(n):
    if n < 0:
            print("Factorial is not defined for negative numbers.")
    elif n == 0 or n == 1:
            print("The factorial of", n, "is 1.")
    else:
            result = 1
            for i in range(2, n + 1):
                result *= i
            return result
n = int(input('enter a number: '))
print('factorial of',n,factorial(n))
```

3.	Create a guessing game where the correct number is 7.

```python
def guessing_game():
    correct_number = 7
    
    while True:
        user = int(input("Pick a number from 1 to 10: "))
        
        if user not in range(1, 11):
            print("Please choose a number between 1 and 10.")
            continue
        
        if user == correct_number:
            print("Your guess was right! 🎉")
            break
        else:
            print("Wrong answer! Try again.")

guessing_game()
```

4.	Check whether a number is prime or not.

```python
def isPrime(num):
    
    if num <= 1:
        print(num,"is not a prime number.")
        return
        
    for i in range(2,num):
        if num % i ==0:
            print(num,"is not a prime number.")
            return
    print(num,"is a prime number.")
print(isPrime(1))
```
