## 🔹 Part 3: String Functions & Indexing

### Q5

Given the string:

```python
text = "python programming"
```

Do the following:

* Print the first character
* Print the last character
* Print the length of the string
* Convert the string to uppercase
* Replace `"python"` with `"java"`

````python
text = 'python programing'

for i in range(len(text)):
    print("Index:", i, "Character:", text[i])

# Print the first character
print('# Print the first character: ' + text[0])

# Print last character
print('# Print the last character: ' + text[-1] + ' or text[16]:' + text[16]) 

#print the lengh of thr STRING
print('# Print the length of the string: ' , len(text))

#Convert the string to uppercase
print('# Convert the string to uppercase: ' + text.upper())

#Replace `"python"` with `"java"`
print('# Replace "python" with "java": ' + text.replace("python", "java"))
````

## 🔹 Part 4: String Slicing

### Q7

Given:

```python
word = "programming"
```

Print:

* `"program"`
* `"ming"`
* The string reversed


````python
word = 'programming'
#print 'program only'
print(word[0:7])
#print 'ming only'
print(word[7:11])
#print reverse of the word
print(word[::-1])
````

## 🔹 Part 5: Lists

### Q9

Create a list of 5 numbers:

* Print the list
* Print the max and min values
* Add a new number to the list
* Remove the last element

````python
lst = [1,2,3,4,5]

#print the element of the list
print(lst)

#print the max value of the list and the min value of the list
print('max value: ',max(lst))
print('min value: ',min(lst))

#add an element to the list
lst.append(6)
print(lst)

#remove the last element of the list NOTE >> pop() method removes the last element of the list and returns it by default
removed_element = lst.pop()
print('removed element:', removed_element)
print(lst)
````
